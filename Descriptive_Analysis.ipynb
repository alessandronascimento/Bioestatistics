# Descriptive Analysis #

In this notebook, we will use different tools to analyze a clinical dataset with stroke data. The dataset is freely available in [HuggingFace](https://huggingface.co/datasets/Nnaodeh/Stroke_Prediction_Dataset).

For this purpose, we will use the following tools:

* [Pandas (Pythond Data Analysis Library)](https://pandas.pydata.org/)
* [Matplotlib](https://matplotlib.org/)
* [Seaborn](https://seaborn.pydata.org/)

They are already installed in Google Colab virtual machines. In case you are running in your local machine, you can install these libraries with pip.

```python
#@title Import libraries and download the dataset
#@markdown Let's inspect the dataset.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("hf://datasets/Nnaodeh/Stroke_Prediction_Dataset/Healthcare-stroke-data.csv")
df
```

## Dataset ##

The dataset has 5,510 entries for 12 columns or categories. The categories are:
1. Id (integer)
2. Gender (Male/Female)
3. Age (float)
4. Hypertension (0/1)
5. Heart disease (0/1)
6. Ever married (Yes/No)
7. Work type (Private/Self-employed/Govt job)
8. Residence type (Rural/Urban)
9. Average glucose level (float)
10. Bmi (body mass index) (float)
11. Smoking status (smokes/never smoked/formely smoked)
12. Stroke (0/1)

Obviously, there are a lot possible questions we can answer looking at this dataset. But before asking questions, let's have an overview of the data.

## Quantitative Variables ##

```python
#@title Histogram for the age
plt.hist(df['age'], bins=15)
plt.xlabel('Age')
plt.ylabel('counts')
plt.title('Age Distribution')
plt.show()
```

```python
#@title Histogram for the average glucose level
plt.hist(df['avg_glucose_level'], bins=15)
plt.xlabel('avg_glucose_level')
plt.ylabel('counts')
plt.title('avg_glucose_level Distribution')
plt.show()
```

```python
#@title Histogram for the average glucose level
plt.hist(df['bmi'], bins=15)
plt.xlabel('bmi')
plt.ylabel('counts')
plt.title('BMI Distribution')
plt.vlines(18, 0 , 1500, color='r')
plt.vlines(25, 0 , 1500, color='r', label='Normal range')
plt.vlines(30, 0 , 1500, color='orange', label='Overweight (Pre-obese)')
plt.legend()
plt.show()
```

## Categorical Data ##

```python
#@title Couting entries by gender
plt.hist(df['gender'])
plt.xlabel('Gender')
plt.ylabel('counts')
plt.title('Gender Distribution')
plt.show()
```

```python
#@title Couting entries by hypertension and heart disease
status = ['No', 'Yes']
plt.bar(status, [len(df[df['hypertension'] == 0]), len(df[df['hypertension'] == 1])], label='Hypertension')
plt.bar(status, [len(df[df['heart_disease'] == 0]), len(df[df['heart_disease'] == 1])], label='heart_disease', alpha=0.6)
plt.xlabel('Status')
plt.ylabel('counts')
plt.legend()
plt.show()
```

# Questions #

# 1 - Are people with high glucose levels more likely to have a stroke? #

Hypothesis: Diabeters increases the chances of have a stroke.

```python
#@title Histogram
plt.hist(df[df['stroke'] == 1]['avg_glucose_level'], label='Stroke')
plt.hist(df[df['stroke'] == 0]['avg_glucose_level'], alpha=0.5, label='No Stroke')
plt.xlabel('Average Glucose Level')
plt.ylabel('Frequency')
plt.title('Average Glucose Level Distribution by Stroke')
plt.legend()
plt.show()
```

```python
#@title Histogram with relative frequencies
plt.hist(df[df['stroke'] == 1]['avg_glucose_level'], label='Stroke', density=True)
plt.hist(df[df['stroke'] == 0]['avg_glucose_level'], alpha=0.5, label='No Stroke', density=True)
plt.xlabel('Average Glucose Level')
plt.ylabel('Frequency')
plt.title('Average Glucose Level Distribution by Stroke')
plt.legend()
plt.show()
```

# Is there any correlation between glucose levels and age ? #

Hypothesis: Diabetes are more frequent among older people.

```python
#@title Scatter plot
plt.scatter(df['age'], df['avg_glucose_level'])
plt.xlabel('Age')
plt.ylabel('Average Glucose Level')
plt.title('Age vs Average Glucose Level')
plt.show()
```

# Is there any association between stroke and age ? #

Hypothesis: Older people are more likely to have a stroke.

```python
#@title Box plot - Ages as a function of stroke status

sns.boxplot(x='stroke', y='age', data=df)
```

```python
#@title Box plot - Glucose levels as a function of the stroke status

sns.boxplot(x='stroke', y='avg_glucose_level', data=df)
```
