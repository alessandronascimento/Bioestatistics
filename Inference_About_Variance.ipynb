The duration of gestation in healthy humans is approximately 280 days with a standard deviation of 10 days.

## 1. What proportion of (healthy) pregnant women will be more than one week “overdue”? Two weeks?

1 week overdue = 287
2 weeks overdue = 294

So, we are looking for

$
P[ Y > x]
$
where x is 287 and 294.

Let's assume a normal distribution for Y.

In this case:

$
Z_{287} = \frac{287-280}{10} = 0.7
$

and

$
Z_{294} = \frac{294-280}{10} = 1.4
$

For $Z > 0.7$

```python
cdf = norm.cdf(0.7, loc=0, scale=1)
print(f'P[Z>{0.7}] = {1-cdf:.4f}')
```

For $Z > 1.4$:

```python
cdf = norm.cdf(1.4, loc=0, scale=1)
print(f'P[Z>{1.4}] = {1-cdf:.4f}')
```

## 2. The gestation periods for a set of four women suffering from a particular condition are 240, 250, 265, and 280 days. Is this evidence that a shorter gestation period is associated with the condition?

We want to compute $\bar{Y}$ and assume that $\mu_{sample} = \bar{Y}$. The hypothesis we want to test is whether $\mu_0 = \mu_{sample}$.

For this purpose, we can compute the limits for $\mu_{sample}$ based on $\bar{Y}, \sigma$ and $n$ and check if these limits include $\mu_0$.

More especifically:

$
\bar{Y} \pm (Z) (\frac{\sigma}{\sqrt{n}})
$



```python
samples = np.array([240, 250, 265, 280])
sample_mean = samples.mean()
print(f'Sample mean: {sample_mean:.2f}')
```

Since the problem states that "Is this evidence that a **shorter** gestation period is associated with the condition", we can use a one-sided test.

For $\alpha=0.05$, $Z=1.64$,

$
(258.75) + (Z)(\sigma/\sqrt{n}) = 258.75 + (1.64)(\sigma/\sqrt{n})
$

```python
deviation = 1.64*10/np.sqrt(4)
upper_limit = sample_mean + deviation
print(f'Lower limit: {upper_limit:.2f}')
```

The upper limit of the sample distribution ($\alpha=0.05$, one-sided) is 267 days, that does not include the populational mean $\mu_0=280$. So, yes. It seems that the condition is associated with a shorter gestation period.

### Where is 258.75 in the population distribution ?

$
Z = \frac{258.75-280}{\sigma/\sqrt{n}} = -4.25
$

So we are 4.25 standard errors off.

```python
z = (258.75-280) / (10/np.sqrt(4))
print(f'Z = {z:.2f}')
```

In terms of probability, we can say it is

```python
cdf = norm.cdf(4.25, loc=0, scale=1)
print(f'P[Z>{4.25}] = {1-cdf:.6f}')
```

or 0.0011%.

## 3. Is the sample variance consistent with the population variance of 102=100? (We assume normality.)

So, we have:

* $\bar{Y} = 258.75$
* $n=3$
* $dof=3$
* $\alpha=0.05$

We are assuming that

$
s^2 \approx \sigma^2
$

This is our *null hypothesis*.

At $\alpha=0.05$ and $dof=3$, we have from the $\chi^2$ table:

```python
alpha = 0.05
degrees_of_freedom = 3
critical_value = stats.chi2.ppf(1-alpha, degrees_of_freedom)
print(f"Critical chi-square value: {critical_value:.4f}")

sample_var = samples.var()
print(f"Sample variance: {sample_var:.4f}")
```

Let's compute our (observed) value of $\chi^2$:

$
\chi^2 = \frac{(n-1)s^2}{\sigma^2} = (3)(229.6875^2) / (100)
$

```python
chi_obs = 3*sample_var**2/100
print(f"Observed Chi square: {chi_obs:.3f}")
```

Since $\chi^2_{obs} \gg \chi^2_{limit}$, we can conclude that the sample variance ($s^2$) is **not** consistent with the population variance ($\sigma^2$).

We can report this result in terms of a p-value as well:

```python
p_value = 1 - stats.chi2.cdf(chi_obs, degrees_of_freedom)
print(f"P-value: {p_value:.8f}")
```

One can read this is as:



> The chances of having this variance from a random sample with N=4 out of the population with variance given by $\mu$ is virtually zero.
