```python
#@title Import libraries
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit
from scipy import stats

# linear model function
def linear_model(x, slope, intercept):
    return intercept + slope * x
```

# Example 1 #

```python
#@title Data
x = np.array([112, 111, 107, 119, 92, 80, 81, 84, 118, 106, 103, 94])
y = np.array([63, 66, 72, 52, 75, 118, 120, 114, 42, 72, 90, 91])

plt.scatter(x, y)
plt.xlabel('Birth weight (ounces)')
plt.ylabel('% Increase in weight between 70-100 days of life')
plt.show()
```

The higher the weight at birth, the **lower** the percentual increase in weight in the 70-100 days of life.

There seems to have a linear relationship between these two variables.

```python
#@title Linear fit

# Perform the curve fit
# popt: optimal values for the parameters (slope, intercept)
# pcov: the estimated covariance of popt
popt, pcov = curve_fit(linear_model, x, y)

slope, intercept = popt

# Compute one standard deviation errors on the parameters
# square root of the diagonal elements of the covariance matrix
perr = np.sqrt(np.diag(pcov))
slope_err, intercept_err = perr

print("--- Fit Results ---")
print(f"Angular coefficient (Slope): {slope:.4f} ± {slope_err:.4f}")
print(f"Linear coefficient (Intercept): {intercept:.4f} ± {intercept_err:.4f}")
```

```python
#@title Plot the data and the fit line

plt.scatter(x, y, label='Data points', color='blue')
x_fit = np.linspace(min(x), max(x), 100)
y_fit = linear_model(x_fit, slope, intercept)
plt.plot(x_fit, y_fit, color='red', label=f'Fit: y = ({slope:.2f}±{slope_err:.2f})x + ({intercept:.2f}±{intercept_err:.2f})')

plt.xlabel('Birth weight (ounces)')
plt.ylabel('% Increase in weight between 70-100 days of life')
plt.legend()
plt.show()
```

## Results ##

Our linear model predicts

$
\hat{Y} = 256 - 1.74(x)
$

as a reasonable model for the data we have. In this context, if the birth weight is 95 oz, it is predicted that the increase between days 70 and 100 of life would be:

$
\hat{y} = 256 - 1.74(95) = 90.7
$

Or and an increase of 90.7% in weight between days 70 and 100 of life.

```python
#@title Computing the Pearson correlation coefficient
r = stats.pearsonr(x, y)
print(f'Pearson correlation coefficient r={r.statistic:.3f} (p={r.pvalue:.5f})')
```

# Example 3 #

A group of $n=9$ kids was studied to assess whether there is a correlation between the pulmonary capacity and the age. Consider the data (below), do the proper test and provide an interpretation for the results given $\alpha=0.05$.

| Age (years) | Pulmonary capacity (L) |
| --- | --- |
|4 |  0.7|
|5 | 0.9 |
|6 | 1.2 |
|7 | 1.3 |
|8 | 1.3 |
|9 | 1.5 |
|10 | 1.7 |
|11 | 1.9 |
|12 | 2.1 |


```python
#@title Data

ex3_x = [4, 5, 6, 7,8, 9, 10, 11, 12]
ex3_y = [0.7, 0.9, 1.2, 1.3, 1.3, 1.5, 1.7, 1.9, 2.1]

plt.scatter(ex3_x, ex3_y)
plt.xlabel('Age (years)')
plt.ylabel('Pulmonary capacity (L)')
plt.show()
```

There seems to be a linear relationship among the data. Let's estimate the regression parameters for this data.

```python
#@title Linear fit

popt, pcov = curve_fit(linear_model, ex3_x, ex3_y)

slope, intercept = popt

perr = np.sqrt(np.diag(pcov))
slope_err, intercept_err = perr

print("--- Fit Results ---")
print(f"Angular coefficient (Slope): {slope:.4f} ± {slope_err:.4f}")
print(f"Linear coefficient (Intercept): {intercept:.4f} ± {intercept_err:.4f}")
print()
print()
print("--- Pearson Correlation Coefficient ---")
r = stats.pearsonr(ex3_x, ex3_y)
print(f'Pearson correlation coefficient r={r.statistic:.3f} (p={r.pvalue:.4f})')
```

```python
#@title Plot the data
plt.scatter(ex3_x, ex3_y, label='Data points', color='blue')
x_fit = np.linspace(min(ex3_x), max(ex3_x), 100)
y_fit = linear_model(x_fit, slope, intercept)
plt.plot(x_fit, y_fit, color='red', label=f'Fit: y = ({slope:.2f}±{slope_err:.2f})x + ({intercept:.2f}±{intercept_err:.2f})')

plt.xlabel('Age (years)')
plt.ylabel('Pulmonary capacity (L)')
plt.legend()
plt.show()
```
