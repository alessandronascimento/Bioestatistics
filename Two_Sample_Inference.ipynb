```python
#@title Import libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy import stats

def q_q_plot_clt(source_data):
    """
    Q-Q plots provide another way to visualize normality of sample means
    """
    fig, axes = plt.subplots(1, 1, figsize=(10, 10))
    # Q-Q plot against standard normal
    stats.probplot(source_data, dist="norm", plot=axes)
    axes.get_lines()[0].set_markerfacecolor('steelblue')
    axes.get_lines()[0].set_markersize(3)

    plt.suptitle(f'Q-Q Plots: Sample Means vs Normal Distribution',fontsize=16)
    plt.tight_layout()
    plt.show()
```

# Normal Distribution

Let's generate a normal distribution centered at $\mu=\pi$ and with $\sigma=0.8$.

First we will generate such distribution using [numpy randn function](https://numpy.org/doc/stable/reference/random/generated/numpy.random.randn.html).

```python
#@markdown $\mu$ = numpy.pi. We have 1,000 points in this distribution.
sigma = 0.8 #@param {'type' : 'number'}

distribution = (np.random.randn(1000)*sigma)+np.pi

fig, axes = plt.subplots(1, 2, figsize=(14, 10))
axes[0].plot(distribution)
axes[0].set_title('Distribution')
axes[0].set_xlabel('Index')
axes[0].set_ylabel('Value')
axes[1].hist(distribution, bins=30)
axes[1].set_title('Histogram')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('Frequency')
plt.show()

print(f'Distribution mean: {distribution.mean():.4f}')
print(f'Distribution std: {distribution.std():.4f}')
```

## Product by a Constant ##

Now, let's see what happens when we multiply the distribution by constants.


```python
a1 = 1.7 #@param {'type' : 'number'}
a2 = 0.7 #@param {'type' : 'number'}

a1_dist = a1*distribution
a2_dist = a2*distribution

print(f'Original distribution mean: {distribution.mean():.4f}')
print(f'Original distribution var: {distribution.var():.4f}')
print(f'Original distribution std: {distribution.std():.4f}')
print()
print(f'Mean for a1 * distribution: {a1_dist.mean():.4f}')
print(f'Variance for a1 * distribution: {a1_dist.var():.4f}')
print(f'Std for a1 * distribution: {a1_dist.std()}')
print()
print(f'Mean for a2 * distribution: {a2_dist.mean():.4f}')
print(f'Variance for a2 * distribution: {a2_dist.var():.4f}')
print(f'Std for a2 * distribution: {a2_dist.std()}')
```

## Distribution of a sum of normally distributed values ###

If $ W = a_1 Y_1 + a_2 Y_2$ with $Y_1$ and $Y_2$ being normally distributed variables, then W is normally distributed.

```python
#@title W = a1_dist + a2_dist
#@markdown W is the sum of the two distributions
W = a1_dist + a2_dist

fig, axes = plt.subplots(3, 2, figsize=(14, 20))
axes[0, 0].plot(a1_dist)
axes[0, 0].set_title('a1 * distribution')
axes[0, 0].set_xlabel('Index')
axes[0, 0].set_ylabel('Value')
axes[0, 1].hist(a1_dist, bins=30)
axes[0, 1].set_title('Histogram')
axes[0, 1].set_xlabel('Value')
axes[0, 1].set_ylabel('Frequency')

axes[1, 0].plot(a2_dist)
axes[1, 0].set_title('a2 * distribution')
axes[1, 0].set_xlabel('Index')
axes[1, 0].set_ylabel('Value')
axes[1, 1].hist(a2_dist, bins=30)
axes[1, 1].set_title('Histogram')
axes[1, 1].set_xlabel('Value')
axes[1, 1].set_ylabel('Frequency')

axes[2, 0].plot(W)
axes[2, 0].set_title('W distribution')
axes[2, 0].set_xlabel('Index')
axes[2, 0].set_ylabel('Value')
axes[2, 1].hist(W, bins=30)
axes[2, 1].set_title('Histogram')
axes[2, 1].set_xlabel('Value')
axes[2, 1].set_ylabel('Frequency')
plt.show()

print(f'Mean for W: {W.mean():.4f}')
print(f'Variance for W: {W.var():.4f}')
print(f'Std for W: {W.std():.4f}')
```

### Is W Normally distributed?

```python
q_q_plot_clt(W)
```

## Another Example ##

Let's work on another example. Here, we will define two distributions:

* $Y_1$ is normally distributed with $\mu_1=100$ and $\sigma_1^2=225$ ($\sigma_1=15$): $Y_1 \sim \mathcal{N}(100,225)$
* $Y_2$ is also normally distributed with $\mu_2=50$ and variance $\sigma_2^2=175$ ($\sigma = 13.23$): $Y_2 \sim \mathcal{N}(50,175)$

From what we have already seen:


|       W         |  E(W)   | Var(W)  |
|-----------------|---------|---------|
| $W_1=Y_1+Y_2$   |   150   |   400   |
| $W_2=Y_1-Y_2$   |   50    |   400   |
| $W_3=2Y_1+Y_2$  |   250   |  1075   |
| $W_4=2Y_1-2Y_2$ |   100   |  1600   |

Let's compute it!

```python
dist_y1 = (np.random.randn(1000)*15)+100
dist_y2 = (np.random.randn(1000)*13.23)+50

W1 = dist_y1 + dist_y2
print(f'W_1 mean = {W1.mean():2f}. W_1 var = {W1.var():2f}')
print()

W2 = dist_y1 - dist_y2
print(f'W_2 mean = {W2.mean():2f}. W_2 var = {W2.var():2f}')
print()

W3 = (2*dist_y1) + dist_y2
print(f'W_3 mean = {W3.mean():2f}. W_3 var = {W3.var():2f}')
print()

W4 = (2*dist_y1) - (2*dist_y2)
print(f'W_4 mean = {W4.mean():2f}. W_4 var = {W4.var():2f}')
print()
```
